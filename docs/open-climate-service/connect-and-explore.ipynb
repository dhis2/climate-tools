This notebook connects to a running [Open Climate Service](https://dhis2.github.io/open-climate-service/) instance, discovers the datasets and workflows it publishes, and opens a dataset as an `xarray.Dataset`.

It needs the `open-climate-service` client (see the [section intro](intro.md)) and the URL of a running instance.

## 1) Connect

Point the client at your instance. No authentication is required for a typical local or country deployment.

```python
from open_climate_service import ClimateService

# Replace with the URL of your Open Climate Service instance
service = ClimateService("https://my-instance.example.org")
```

## 2) Discover published datasets

`datasets()` returns the published collections from the instance's STAC catalog. Each entry has at least an `id`, a `title`, and an `href`.

```python
datasets = service.datasets()
for d in datasets:
    print(d["id"], "—", d.get("title", ""))
```

## 3) See the available processing

An instance exposes both the standard [openEO](https://openeo.org/) processes and any reusable **workflows** (stored process graphs) it ships — such as the org-unit aggregation workflows used later in this section.

```python
# Reusable, named workflows (e.g. aggregate_to_dhis2_org_units)
print("Workflows:")
for wf in service.workflows():
    print(" -", wf["id"])

# Standard openEO processes (mean, sum, filter_temporal, ...)
print(f"\n{len(service.processes())} openEO processes available")
```

## 4) Open a dataset as xarray

`open_dataset()` reads the published GeoZarr store directly over HTTP and returns a lazy `xarray.Dataset` (this step needs the `[xarray]` extra). Nothing is downloaded until you compute or plot.

```python
# Open the first published dataset (or pass any id from the list above)
ds = service.open_dataset(datasets[0]["id"])
ds
```

Because the data is lazy, subset in space and time *before* calling `.compute()` or plotting — see the [Data visualization](../guides/visualization/intro.md) guide for examples.

```python
variable = list(ds.data_vars)[0]
ds[variable]
```
